```python
import pandas as pd
import glob
import os

path = r"C:\Users\anitc\Downloads\python hackathon\HUPA-UC Diabetes Dataset\*.csv"

files = glob.glob(path)
print("Files found:", len(files))

dfs = []
for f in files:
    print("Loading:", f)
    df = pd.read_csv(f, encoding="latin1")  # handles special characters
    df["participant_id"] = os.path.splitext(os.path.basename(f))[0]
    dfs.append(df)

data = pd.concat(dfs, ignore_index=True)
print("Combined shape:", data.shape)
```
